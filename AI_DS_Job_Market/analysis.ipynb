```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("AI_Job_Market_Trends_2026.csv")
df.head()
```

```python
df.info()
df.isnull().sum()
```

```python
df['job_title'].value_counts().head(10)
```

```python
df['job_title'].value_counts().head(10).plot(kind='bar', figsize=(10,5))
plt.title("Top 10 Job Roles")
plt.xlabel("Job Title")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

Insight:

Data Analyst and AI Engineer roles are in the highest demand.

Business Analyst is also a high-demand role.

```python
sns.countplot(x='company_size', data=df)
plt.title("Company Size Distribution")
plt.show()
```

Insight:

MNCs and startups are hiring more.

```python
df['country'].value_counts().head(10).plot(kind='bar')
plt.title("Top Hiring Countries")
plt.show()
```

Insight:

Germany, Singapore top hiring countries

```python
sns.countplot(x='remote_type', data=df)
plt.title("Remote vs Onsite Jobs")
plt.show()
```

Insight:

The trend of remote jobs is increasing.

```python
df['company_industry'].value_counts().head(10).plot(kind='bar')
plt.title("Top Industries Hiring")
plt.show()
```

Insight:

Technology & E-commerce leading industries

```python
pd.crosstab(df['job_title'], df['company_size']).head()
```

Insight:

Different types of companies hire for different roles depending on their needs and domain.
